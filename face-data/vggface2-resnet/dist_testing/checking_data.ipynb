```python
dim_0 = user_1['dim_0'].to_numpy()
```

```python
stat, p = shapiro(dim_0)
print(f"Shapiro-Wilk p-value: {p}")
if p > 0.05: print("Probably Gaussian")
else: print("Probably NOT Gaussian")
```

```python
def count_gaussian_dims(df: pl.DataFrame, threshold: int = 20, max_dims: int = 25):
    keys = df["KEY"].unique().to_list()
    results = []
    dim_cols = [f"dim_{i}" for i in range(min(max_dims, len(df.columns) - 1))]
    
    for key in tqdm(keys):
        sub_df = df.filter(pl.col("KEY") == key)
        
        if sub_df.height < threshold: continue

        count = 0
        for col in dim_cols:
            data = sub_df[col].to_numpy()

            if len(data) >= 3:
                try:
                    _, p = shapiro(data)
                    if p > 0.05: count += 1
                except Exception: pass  
        results.append((key, count))
    return pl.DataFrame(results, schema=["KEY", "VALID"])
```

```python
gauss_check_df = count_gaussian_dims(df)
```

```python
print(gauss_check_df['VALID'].describe())
```

```python
gauss_check_df_25 = count_gaussian_dims(df, max_dims = 75)
print(gauss_check_df_25['VALID'].describe())
```
